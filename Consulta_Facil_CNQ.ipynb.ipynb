### \*CONTRUÇÃO BANCO DE DADOS\*

```
# Ativar a verificação dos relacionamentos
cursor.execute("PRAGMA foreign_keys = ON")

# 1. Tabela de áreas de formação
cursor.execute("""
CREATE TABLE IF NOT EXISTS areas_formacao (
    codigo_area TEXT PRIMARY KEY,
    designacao_area TEXT
)
""")

# 2. Tabela de qualificações
cursor.execute("""
CREATE TABLE IF NOT EXISTS qualificacoes (
    codigo_qualificacao TEXT PRIMARY KEY,
    qualificacao TEXT,
    codigo_area TEXT,
    nivel_qnq TEXT,
    nivel_qeq TEXT,
    FOREIGN KEY (codigo_area)
        REFERENCES areas_formacao(codigo_area)
)
""")

# 3. Tabela de UFCDs
cursor.execute("""
CREATE TABLE IF NOT EXISTS ufcd (
    codigo_ufcd TEXT PRIMARY KEY,
    designacao TEXT,
    carga_horaria TEXT
)
""")

# 4. Tabela de associações
cursor.execute("""
CREATE TABLE IF NOT EXISTS associacoes (
    id_associacao INTEGER PRIMARY KEY AUTOINCREMENT,
    codigo_ufcd TEXT NOT NULL,
    codigo_qualificacao TEXT,
    componente TEXT,
    FOREIGN KEY (codigo_ufcd)
        REFERENCES ufcd(codigo_ufcd),
    FOREIGN KEY (codigo_qualificacao)
        REFERENCES qualificacoes(codigo_qualificacao)
)
""")

conexao.commit()

print("As quatro tabelas foram criadas!")
```

# Conferir as áreas de formação

```
resumo_areas = df.groupby(
    "Código Área de Formação"
).agg(
    quantidade_designacoes=(
        "Designação Área de Formação", "nunique"
    )
).reset_index()

conflitos_areas = resumo_areas[
    resumo_areas["quantidade_designacoes"] > 1
]

print("Áreas de formação diferentes:",
      df["Código Área de Formação"].nunique())

print("Códigos de área com divergências:",
      len(conflitos_areas))

display(conflitos_areas)
```

# Inserir os dados nas tabelas

```
# 1. Preparar as áreas de formação
areas = df[
    ["Código Área de Formação", "Designação Área de Formação"]
].dropna(subset=["Código Área de Formação"])

areas = areas.drop_duplicates(
    subset=["Código Área de Formação"]
).rename(columns={
    "Código Área de Formação": "codigo_area",
    "Designação Área de Formação": "designacao_area"
})

# 2. Preparar as qualificações
qualificacoes = df[
    [
        "Código Qualificação",
        "Qualificação",
        "Código Área de Formação",
        "Nível QNQ",
        "Nível QEQ"
    ]
].dropna(subset=["Código Qualificação"])

qualificacoes = qualificacoes.drop_duplicates(
    subset=["Código Qualificação"]
).rename(columns={
    "Código Qualificação": "codigo_qualificacao",
    "Qualificação": "qualificacao",
    "Código Área de Formação": "codigo_area",
    "Nível QNQ": "nivel_qnq",
    "Nível QEQ": "nivel_qeq"
})

# 3. Preparar as UFCDs
ufcds = df[
    [
        "Código UFCD",
        "UFCD",
        "Carga Horária UFCD"
    ]
].drop_duplicates(
    subset=["Código UFCD"]
).rename(columns={
    "Código UFCD": "codigo_ufcd",
    "UFCD": "designacao",
    "Carga Horária UFCD": "carga_horaria"
})

# 4. Preparar todas as associações
associacoes = df[
    [
        "Código UFCD",
        "Código Qualificação",
        "Componente"
    ]
].copy().rename(columns={
    "Código UFCD": "codigo_ufcd",
    "Código Qualificação": "codigo_qualificacao",
    "Componente": "componente"
})

# 5. Inserir nas tabelas SQLite
areas.to_sql("areas_formacao", conexao, if_exists="append", index=False)

qualificacoes.to_sql(
    "qualificacoes", conexao, if_exists="append", index=False
)

ufcds.to_sql("ufcd", conexao, if_exists="append", index=False)

associacoes.to_sql(
    "associacoes", conexao, if_exists="append", index=False
)

conexao.commit()

print("Dados inseridos com sucesso!")
```

## \*Primeira consulta SQL\*

```
consulta_sql = """
SELECT
    u.codigo_ufcd,
    u.designacao AS nome_ufcd,
    u.carga_horaria,
    a.componente,
    q.codigo_qualificacao,
    q.qualificacao,
    q.nivel_qnq,
    q.nivel_qeq
FROM ufcd u

LEFT JOIN associacoes a
    ON u.codigo_ufcd = a.codigo_ufcd

LEFT JOIN qualificacoes q
    ON a.codigo_qualificacao = q.codigo_qualificacao

WHERE u.codigo_ufcd = ?

ORDER BY q.qualificacao
"""

resultado = pd.read_sql_query(
    consulta_sql,
    conexao,
    params=("0349",)
)

display(resultado)
```

## Criar uma pesquisa por código ou nome

```
termo = input("Digite o código ou nome da UFCD: ")

consulta_sql = """
SELECT DISTINCT
    u.codigo_ufcd,
    u.designacao,
    u.carga_horaria,
    a.componente,
    q.codigo_qualificacao,
    q.qualificacao,
    q.nivel_qnq,
    q.nivel_qeq
FROM ufcd u

LEFT JOIN associacoes a
    ON u.codigo_ufcd = a.codigo_ufcd

LEFT JOIN qualificacoes q
    ON a.codigo_qualificacao = q.codigo_qualificacao

WHERE u.codigo_ufcd LIKE ?
   OR u.designacao LIKE ?

ORDER BY u.codigo_ufcd
"""

resultado = pd.read_sql_query(
    consulta_sql,
    conexao,
    params=(f"%{termo}%", f"%{termo}%")
)

if resultado.empty:
    print("Nenhuma UFCD encontrada.")
else:
    display(resultado)
```

```

import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin

paginas = {
    "UFCD": "https://catalogo.snq.gov.pt/ufcdPesquisa",
    "UC": "https://catalogo.snq.gov.pt/ucPesquisa",
    "Qualificações": "https://catalogo.snq.gov.pt/qualificacoesPesquisa"
}

for nome, url in paginas.items():

    resposta = requests.get(url, timeout=30)
    soup = BeautifulSoup(resposta.text, "html.parser")

    print("\n" + "=" * 40)
    print(nome)
    print("=" * 40)

    encontrados = 0

    for link in soup.find_all("a", href=True):
        texto = link.get_text(" ", strip=True)

        if "descarregar" in texto.lower() or "download" in texto.lower():
            print("Texto:", texto)
            print("Link:", urljoin(url, link["href"]))
            encontrados += 1

    for formulario in soup.find_all("form"):
        print("Formulário:", formulario.get("action"))
        print("Método:", formulario.get("method"))

    print("Links de download encontrados:", encontrados)
```

# Criar a atualização segura

```

import pandas as pd
import sqlite3
from pathlib import Path
from datetime import datetime

# Arquivos de origem
arquivos = {
    "dados_originais": {
        "arquivo": "ListaUFCDs_2026-10-30.xlsx",
        "aba": "UFCD",
        "codigo": "Código UFCD"
    },
    "ucs_originais": {
        "arquivo": "ListaUCs_2026-10-30.xlsx",
        "aba": "UC",
        "codigo": "Código UC"
    },
    "qualificacoes_lista": {
        "arquivo": "ListaQualificacoes_2026-10-31.xlsx",
        "aba": "Qualificações",
        "codigo": "Código Qualificação"
    }
}

banco = Path("consulta_facil_cnq.db")

if not banco.exists():
    raise FileNotFoundError("Banco de dados não encontrado.")

# 1. Ler e validar os três arquivos antes de alterar o banco
dados = {}

for tabela, info in arquivos.items():
    caminho = Path(info["arquivo"])

    if not caminho.exists():
        raise FileNotFoundError(f"Arquivo não encontrado: {caminho}")

    df = pd.read_excel(
        caminho,
        sheet_name=info["aba"],
        skiprows=2,
        dtype=str
    )

    if tabela == "qualificacoes_lista":
        df = df.drop(columns=["Unnamed: 0"], errors="ignore")

    if df.empty:
        raise ValueError(f"A lista {tabela} está vazia.")

    if info["codigo"] not in df.columns:
        raise ValueError(f"Coluna de código ausente em {tabela}.")

    if df[info["codigo"]].isna().any():
        raise ValueError(f"Existem códigos vazios em {tabela}.")

    dados[tabela] = df

# 2. Criar backup consistente do banco
data = datetime.now().strftime("%Y%m%d_%H%M%S")
backup = Path(f"consulta_facil_cnq_backup_{data}.db")

origem = sqlite3.connect(str(banco))
destino = sqlite3.connect(str(backup))

try:
    origem.backup(destino)
finally:
    destino.close()
    origem.close()

print("Backup criado:", backup.name)

# 3. Preparar tabelas temporárias
conexao = sqlite3.connect(str(banco))

temporarias = {
    "dados_originais": "temp_dados_originais",
    "ucs_originais": "temp_ucs_originais",
    "qualificacoes_lista": "temp_qualificacoes_lista"
}

try:
    for tabela, df in dados.items():
        df.to_sql(
            temporarias[tabela],
            conexao,
            if_exists="replace",
            index=False
        )

    # 4. Substituir as três tabelas numa única transação
    conexao.execute("BEGIN IMMEDIATE")

    for tabela, temporaria in temporarias.items():
        total = conexao.execute(
            f'SELECT COUNT(*) FROM "{temporaria}"'
        ).fetchone()[0]

        if total != len(dados[tabela]):
            raise ValueError(f"Quantidade incorreta na tabela {tabela}.")

        conexao.execute(f'DROP TABLE IF EXISTS "{tabela}"')
        conexao.execute(
            f'ALTER TABLE "{temporaria}" RENAME TO "{tabela}"'
        )

    conexao.commit()

    print("\nATUALIZAÇÃO CONCLUÍDA COM SUCESSO!")

    for tabela, df in dados.items():
        print(f"{tabela}: {len(df)} registros")

except Exception:
    conexao.rollback()
    raise

finally:
    conexao.close()
```
